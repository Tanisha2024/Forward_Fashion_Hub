```python
import pandas as pd
faq=pd.read_csv("../data/faq.csv")
products=pd.read_csv("../data/products.csv")
print('faq',faq.shape)
print('products',products.shape)

```

```python
faq_documents=[]
for _, row in faq.iterrows():
    documnet=f"""
Question:{row['question']}
Answer:{row['answer']}
"""
    faq_documents.append(documnet.strip())
print(faq_documents[0])
```

```python
product_documents=[]
for _,row in products.iterrows():
    documnet=f"""
Product:{row['name']}
Category:{row['category']}
Descripion:{row['description']}
Price:₹{row['price']}
Color:{row['color']}
Available sizes:{row['size'].replace('|',',')}
Material:{row['material']}
Occasion:{row['occasion'].replace('|',',')}
Style:{row['style']}
"""
    product_documents.append(documnet.strip())
print(product_documents[0])
```

```python
product_metadata=[]
for _, row in products.iterrows():
    metadata={
        "product_id":row['product_id'],
        "category":row['category'],
        "price":row['price'],
        "color":row['color'],
        "material":row['material'],
        "occasion":row['occasion'],
        "style":row['style']
    }
    product_metadata.append(metadata)
print(product_metadata[0])
```
